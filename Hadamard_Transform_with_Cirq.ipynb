# Measurements After a Hadamard Transform on a Qubit Using Cirq
## Study notes by Gabriel Velazquez

```python
import cirq

# Define a Qubit
qubit = cirq.GridQubit(0, 0)

# Create a circuit in cirq
circuit = cirq.Circuit(
    cirq.H(qubit),
    cirq.measure(qubit, key='m')
)

print("Circuit Follows")
print(circuit)

# Initialize the simulator
sim = cirq.Simulator()

# Execute the circuit n times
n = 100
output = sim.run(circuit, repetitions=n)

print("\nMeasurement Output:")
print(output)

print("\nHistogram stats follow")
print(output.histogram(key='m'))
```
