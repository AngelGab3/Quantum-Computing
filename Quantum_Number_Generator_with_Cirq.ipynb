```python
import cirq
import numpy as np

def random_number_generator(low=0, high=2**10, m=10):
    """
    :param low: lower bound of numbers to be generated
    :param high: Upper bound of numbers to be generated
    :param number m : Number of random numbers to output
    :return: list of random numbers
    """
    # Determine the number of Qubits required
    qubits_required = int(np.ceil(np.log2(high - low)))
    print(f"Qubits required: {qubits_required}")

    Q_reg = cirq.LineQubit.range(qubits_required)

    circuit = cirq.Circuit()

    circuit.append(cirq.H.on_each(*Q_reg))

    circuit.append(cirq.measure(*Q_reg, key="z"))

    print("\nCircuit Follows:")
    print(circuit)

    sim = cirq.Simulator()

    num_gen = 0
    output = []

    while num_gen < m:
        result = sim.run(circuit, repetitions=1)
        bitstring = result.measurements['z'][0]
        measured_int = int("".join(map(str, bitstring)), 2)

        rand_number = measured_int + low

        if rand_number < high:
            output.append(rand_number)
            num_gen += 1

    return output

if __name__ == '__main__':
    output = random_number_generator()
    print("\nGenerated Random Numbers:")
    print(output)
    print(f"Mean of generated numbers: {np.mean(output)}")
```
