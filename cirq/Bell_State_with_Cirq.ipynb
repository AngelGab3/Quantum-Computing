```python
import cirq
q_register = cirq.LineQubit.range(2)
circuit = cirq.Circuit(
    cirq.H(q_register[0]),
    cirq.CNOT(q_register[0], q_register[1])
)
circuit.append(cirq.measure(*q_register, key='z'))

print("Circuit Follows:")
print(circuit)

sim = cirq.Simulator()
output = sim.run(circuit, repetitions=100)

print("\nMeasurement Output:")
print(output.histogram(key='z'))
```
